# 2.1. 로봇의 좌표

```python
import sympy as sp
import numpy as np

p = sp.symbols('P')
i, j, k = sp.symbols('i j k')

P = sp.Matrix([p*i, p*j, p*k]).T
print(P)
```

# 2.2. 기본 행렬과 벡터

```python
# 행렬곱
x1, x2, a11, a12, a21, a22 = sp.symbols(
    'x1 x2 a11 a12 a21 a22'
)

A = np.array([
    [a11, a12],
    [a21, a21]
])
x = np.array([x1, x2])

y = A@x
y
```

```python
# 단위벡터
a, b, i, j, k = sp.symbols('a b i j k')

a_vec = np.array([a*i, a*j, a*k])
b_vec = np.array([b*i, b*j, b*k])

c = a_vec + b_vec
sp.simplify(c)
```
