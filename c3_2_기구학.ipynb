```python
from sympy import symbols, sin, cos, trigsimp
import numpy as np

t1 = symbols('theta1')
t2 = symbols('theta2')
l1 = symbols('l1'); l2 = symbols('l2')
c1 = cos(t1); s1 = sin(t1)
c2 = cos(t2); s2 = sin(t2)

A01 = np.array([
    [c1, -s1, 0, l1*c1],
    [s1, c1, 0, l1*s1],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

A12 = np.array([
    [c2, -s2, 0, l2*c2],
    [s2, c2, 0, l2*s2],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])
#print(t1, c1, c2, l1, l2, s1, s2)
A02 = trigsimp(A01 @ A12)

for i in A02:
    print(i)
```

```python
from numpy import cos, sin, pi
import numpy as np

def solve(t1, t2, l1, l2):
    c1 = cos(t1); c2 = cos(t2)
    s1 = sin(t1); s2 = sin(t2)
    A01 = np.array([
        [c1, -s1, 0, l1*c1],
        [s1, c1, 0, l1*s1],
        [0, 0, 1, 0],
        [0, 0, 0, 1]
    ])

    A12 = np.array([
        [c2, -s2, 0, l2*c2],
        [s2, c2, 0, l2*s2],
        [0, 0, 1, 0],
        [0, 0, 0, 1]
    ])

    A02 = A01 @ A12
    return A02

print(solve(pi/6, pi/6, 0.432, 0.432))
(solve(pi/4, pi/4, 0.432, 0.432))
```
