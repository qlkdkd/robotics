```python
# mat ex3.2
from numpy import cos, sin, pi
import numpy as np
import matplotlib.pyplot as plt

def solve(t1, t2, l1, l2):
    c1 = cos(t1); c2 = cos(t2)
    s1 = sin(t1); s2 = sin(t2)
    A01 = np.array([
        [c1, -s1, 0, l1*c1],
        [s1, c1, 0, l1*s1],
        [0, 0, 1, 0],
        [0, 0, 0, 1]
    ])

    A12 = np.array([
        [c2, -s2, 0, l2*c2],
        [s2, c2, 0, l2*s2],
        [0, 0, 1, 0],
        [0, 0, 0, 1]
    ])

    A02 = A01 @ A12
    return A02

a_30 = solve(pi/6, pi/6, 0.432, 0.432)
a_45 = solve(pi/4, pi/4, 0.432, 0.432)

print(a_30)
print(a_45)

l1 = 0.432; l2 = 0.432
for i in range(1, 4):
    t2 = pi / 6 * i; t1 = pi / 6 * i
    x1 = l1 * cos(t1); y1 = l1 * sin(t1)

    x = l1 * cos(t1) + l2 * cos(t1 + t2)
    y = l1 * sin(t1) + l2 * sin(t1 + t2)

    rx = [0, x1, x]
    ry = [0, y1, y]
    plt.plot(rx, ry, '-')

plt.axis([-1, 1, -1, 1])
plt.gca().set_aspect('equal', adjustable='box')
plt.title('2 link robot')
plt.xlabel('x axis (m)')
plt.ylabel('y axis (m)')
plt.grid(True)
```

```python
# ex 3.2

from sympy import symbols, sin, cos, trigsimp
import numpy as np

l1 = symbols('l1'); l2 = symbols('l2'); l3 = symbols('l3')
t1 = symbols('t1'); t2 = symbols('t2'); t3 = symbols('t3')
c1 = cos(t1); c2 = cos(t2); c3 = cos(t3)
s1 = sin(t1); s2 = sin(t2); s3 = sin(t3)

A01 = np.array([
    [c1, -s1, 0, l1*c1],
    [s1, c1, 0, l1*s1],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

A12 = np.array([
    [c2, -s2, 0, l2*c2],
    [s2, c2, 0, l2*s2],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

A23 = np.array([
    [c3, -s3, 0, l3*c3],
    [s3, c3, 0, l3*s3],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

T03 = trigsimp(A01 @ A12 @ A23)
print('T03')
for i in T03:
    print(i)
```

```python
from sympy import symbols, sin, cos, trigsimp
import numpy as np

d1 = symbols('d1'); d2 = symbols('d2'); d3 = symbols('d3')
t1 = symbols('t1'); t2 = symbols('t2'); t3 = symbols('t3')
c1 = cos(t1); c2 = cos(t2); c3 = cos(t3)
s1 = sin(t1); s2 = sin(t2); s3 = sin(t3)

A01 = np.array([
    [c1, 0, s1, 0],
    [s1, 0, -c1, 0],
    [0, 0, 1, d1],
    [0, 0, 0, 1]
])

A12 = np.array([
    [c2, -s2, 0, d2*c2],
    [s2, c2, 0, d2*s2],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

A23 = np.array([
    [c3, -s3, 0, d3*c3],
    [s3, c3, 0, d3*s3],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

T03 = trigsimp(A01 @ A12 @ A23)
print('T03')
for i in T03:
    print(i)
```

```python
# ex3.4
from numpy import cos, sin, pi, sqrt, asin, acos
import matplotlib.pyplot as plt

l1 = 1; l2 = 1; l3 = 1; l4 = 1

for i in range(0, 40, 4):
    theta1 = pi/3 + pi/2 - pi/60 * i
    s = sqrt(l1**2 + l2**2 + 2*l1*l2*cos(theta1))
    beta = asin(l2/s * sin(theta1))
    psi = acos((l3**2 + s**2 - l4**2) / (2*l3*3))
    lambda_ = asin(l3 / l4 * sin(psi))
    theta2 = psi - beta
    theta3 = 2 * pi - (lambda_ + beta)

    r1x = l2 * cos(theta1)
    r1y = l2 * sin(theta1)
    r2x = l2 * cos(theta1) + l3 * cos(theta2)
    r2y = l2 * sin(theta1) + l3 * cos(theta2)
    r3x = r2x + l4 * cos(beta + lambda_)
    r3y = r2y - l4 * sin(beta + lambda_)

    x = l2 * cos(theta1) + l3 * cos(theta2)
    y = l2 * sin(theta1) + l3 * sin(theta2) 

    rx = [0, r1x, r2x, r3x]
    ry = [0, r1y, r2y, r3y]

    plt.plot(rx, ry)
    plt.scatter(x, y)

plt.show()
```
